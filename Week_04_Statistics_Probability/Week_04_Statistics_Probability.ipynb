## Importing the libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

## 1. Mean, Median, and Mode

These are measures of central tendency.
- Mean: the average value.
- Median: the middle value when the data is ordered.
- Mode: the value that appears most frequently.

These measures help describe the center of a dataset.

```python
data = np.array([10, 20, 20, 30, 40, 50, 50, 50, 60])

mean_value = np.mean(data)
median_value = np.median(data)

mode_value = pd.Series(data).mode()[0]

print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

## 2. Variance

Variance measures how much the values in a dataset differ from the mean. A larger variance means the observations are more spread out. For this exercise, we use population variance with NumPy's default `ddof=0`.

```python
variance_value = np.var(data)

print("Variance:", variance_value)
```

## 3. Standard Deviation

Standard deviation describes the typical spread of values around the mean. It is the square root of variance. A smaller standard deviation means the values are generally closer to the mean.

```python
std_value = np.std(data)

print("Standard deviation:", std_value)
```

## 4. Correlation

Correlation measures the strength and direction of a linear relationship between two variables. The correlation coefficient ranges from -1 to +1.
- A value close to +1 indicates a strong positive linear relationship.
- A value close to -1 indicates a strong negative linear relationship.
- A value close to 0 indicates little or no linear relationship.

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation:", correlation)
```

## 5. Basic Probability

Probability describes how likely an event is to occur.The probability of an event ranges from 0 to 1.
- 0 means the event is impossible.
- 1 means the event is certain.
- Values between 0 and 1 represent different levels of likelihood.
For example, when rolling a fair six-sided die, the probability of getting a 6 is:P(6) = 1 / 6

```python
probability_of_six = 1 / 6

print("Probability of rolling a 6:", probability_of_six)
```

## 6. Conditional Probability

Conditional probability describes the probability of an event occurring given that another event has already occurred.


```python
passed_students = 12
passed_and_completed = 8

conditional_probability = passed_and_completed / passed_students

print(
    "P(Completed | Passed):",
    conditional_probability
)
```

## 7. Normal Distribution

The normal distribution is a common probability distribution with a symmetric, bell-shaped form. here, we will create a simple simulated normal distribution to understand its shape.


```python
np.random.seed(42)

normal_data = np.random.normal(
    loc=50,
    scale=10,
    size=1000
)

print("Mean:", normal_data.mean())
print("Standard deviation:", normal_data.std())
```

```python
plt.hist(normal_data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Simulated Normal Distribution")
plt.show()
```

## 8. Central Limit Theorem

The Central Limit Theorem (CLT) describes an important property of sample means. When we repeatedly take sufficiently large random samples from a population, the distribution of the sample means tends to become approximately normal, even when the original population is not normally distributed. This is important because it allows statistical methods based on normal distributions to be used with many types of real-world data.

```python
np.random.seed(42)

population = np.random.exponential(
    scale=10,
    size=10000
)

sample_means = []

for _ in range(1000):
    sample = np.random.choice(
        population,
        size=30,
        replace=True
    )
    sample_means.append(sample.mean())

sample_means = np.array(sample_means)

print("Mean of sample means:", sample_means.mean())
print("Standard deviation of sample means:", sample_means.std())
```

```python
plt.hist(sample_means, bins=30)
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.title("Distribution of Sample Means")
plt.show()
```

## Central Limit Theorem – Interpretation

The Central Limit Theorem states that when we repeatedly take sufficiently large random samples from a population, the distribution of the sample means tends to become approximately normal, even when the original population is not normally distributed. In this example, the original population was generated from an exponential distribution, which is not normally distributed. However, the histogram of the sample means is approximately bell-shaped. The mean of the sample means is approximately 9.84, while the standard deviation is approximately 1.72. This demonstrates the Central Limit Theorem at an intuitive level.

## 5. Data Cleaning

The dataset contains no missing values and no duplicate rows. The numerical score columns are already stored as integers. For consistency and easier analysis, I will standardize the column names by converting them to lowercase and replacing spaces with underscores.

```python
# Create a copy of the raw dataset
cleaned_df = raw_df.copy()

# Standardize column names
cleaned_df.columns = (
    cleaned_df.columns
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("/", "_")
)

# Display the cleaned column names
print("Cleaned column names:")
print(cleaned_df.columns.tolist())
```

## 6. Descriptive Statistics

Descriptive statistics help summarize the main characteristics of numerical data.
For the student scores, I will calculate: Mean, Median, Mode, Variance, Standard deviation. These statistics will help describe the distribution and variation of math, reading, and writing scores.

```python
# Select the numerical score columns
score_columns = ["math_score", "reading_score", "writing_score"]

# Calculate mean, median, and mode
for column in score_columns:
    mean_value = cleaned_df[column].mean()
    median_value = cleaned_df[column].median()
    mode_value = cleaned_df[column].mode()[0]

    print(f"\n{column}")
    print("Mean:", mean_value)
    print("Median:", median_value)
    print("Mode:", mode_value)
```

### Variance and Standard Deviation

Variance measures how much the scores vary around their mean. Standard deviation is the square root of variance and provides a measure of the typical spread of the scores around the mean. I will calculate both measures for the three score variables.

```python
# Calculate variance and standard deviation
for column in score_columns:
    variance_value = cleaned_df[column].var()
    std_value = cleaned_df[column].std()

    print(f"\n{column}")
    print("Variance:", variance_value)
    print("Standard deviation:", std_value)
```

## 7. Correlation Analysis

Correlation measures the strength and direction of a linear relationship between two numerical variables.
I will examine the relationships between:

- Math and reading scores
- Math and writing scores
- Reading and writing scores
A positive correlation means that higher values of one variable tend to be associated with higher values of the other variable.

```python
# Calculate the correlation matrix for the three score variables
correlation_matrix = cleaned_df[score_columns].corr()

print("Correlation Matrix:")
print(correlation_matrix)
```

## 7. Correlation Analysis

Correlation measures the strength and direction of a linear relationship between two numerical variables.

```python
# Calculate the correlation matrix for the three score variables
correlation_matrix = cleaned_df[score_columns].corr()

print("Correlation Matrix:")
print(correlation_matrix)
```

### Correlation Heatmap

A heatmap provides a visual representation of the correlation matrix. Values closer to +1 indicate a strong positive linear relationship, while values closer to 0 indicate a weaker linear relationship.

```python
plt.figure(figsize=(8, 6))

plt.imshow(correlation_matrix, cmap="coolwarm", vmin=-1, vmax=1)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(score_columns)),
    score_columns,
    rotation=45
)

plt.yticks(
    range(len(score_columns)),
    score_columns
)

plt.title("Correlation Heatmap of Student Scores")

# Display correlation values inside the heatmap
for i in range(len(score_columns)):
    for j in range(len(score_columns)):
        plt.text(
            j,
            i,
            f"{correlation_matrix.iloc[i, j]:.2f}",
            ha="center",
            va="center",
            color="black"
        )

plt.tight_layout()
plt.show()
```
