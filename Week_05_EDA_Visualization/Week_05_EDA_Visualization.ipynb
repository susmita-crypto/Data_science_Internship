## 4. Histogram – Passenger Age

A histogram is used to examine the distribution of passenger ages.

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Age"], bins=20)

plt.title("Distribution of Passenger Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

### Interpretation

The histogram shows how passenger ages are distributed across different age ranges and helps identify the ranges containing more passengers.

## 5. Box Plot – Fare by Passenger Class

A box plot is used to examine the distribution and spread of ticket fares across passenger classes.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Pclass", y="Fare")

plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")

plt.show()
```

### Interpretation

The box plot compares the distribution of ticket fares across passenger classes and shows differences in their spread and typical values.

## 6. Scatter Plot – Age and Fare

A scatter plot is used to examine the relationship between passenger age and ticket fare.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived"
)

plt.title("Relationship Between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

### Interpretation

Each point represents a passenger. The chart shows how age and fare are distributed together, while the grouping indicates survival status.

## 7. Correlation Heatmap

A correlation heatmap is used to examine relationships between numerical variables.

Correlation values range from -1 to 1. Positive values indicate a positive linear relationship, while negative values indicate a negative linear relationship.

```python
numeric_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare"
]

correlation_matrix = df[numeric_columns].corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

## 8. Survival Rate by Passenger Class

The average survival rate is calculated for each passenger class to identify differences between the groups.

```python
survival_by_class = df.groupby("Pclass")["Survived"].mean()

print(survival_by_class)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=survival_by_class.index,
    y=survival_by_class.values
)

plt.title("Average Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Survival Rate")

plt.show()
```

## 9. Survival Rate by Sex

The average survival rate is calculated by sex to compare the groups.

```python
survival_by_sex = df.groupby("Sex")["Survived"].mean()

print(survival_by_sex)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=survival_by_sex.index,
    y=survival_by_sex.values
)

plt.title("Average Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Average Survival Rate")

plt.show()
```
