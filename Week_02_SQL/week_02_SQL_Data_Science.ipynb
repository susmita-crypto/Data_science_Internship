## 1. Create SQLite Database

SQLite is used to store and query our sample sales data.

```python
conn = sqlite3.connect("sales_database.db")

print("Database connected successfully!")
```

## 2. Create Tables
here to create the table i will use four tables customers, products, orders, order_items

```python
conn.executescript("""
CREATE TABLE IF NOT EXISTS customers (
    customer_id INTEGER PRIMARY KEY,
    customer_name TEXT,
    city TEXT,
    age INTEGER
);

CREATE TABLE IF NOT EXISTS products (
    product_id INTEGER PRIMARY KEY,
    product_name TEXT,
    category TEXT,
    price REAL
);

CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    order_date TEXT
);

CREATE TABLE IF NOT EXISTS order_items (
    order_item_id INTEGER PRIMARY KEY,
    order_id INTEGER,
    product_id INTEGER,
    quantity INTEGER
);
""")

print("Tables created successfully!")
```

## 3. Insert Sample Data

```python
customers = [
    (1, "Ram", "Kathmandu", 22),
    (2, "Sita", "Pokhara", 25),
    (3, "Hari", "Janakpur", 21),
    (4, "Anita", "Kathmandu", 28),
    (5, "Bikash", "Biratnagar", 30)
]

products = [
    (1, "Laptop", "Electronics", 80000),
    (2, "Mouse", "Electronics", 1500),
    (3, "Keyboard", "Electronics", 3000),
    (4, "Chair", "Furniture", 15000),
    (5, "Notebook", "Stationery", 200)
]

orders = [
    (101, 1, "2026-09-01"),
    (102, 2, "2026-09-02"),
    (103, 3, "2026-09-03"),
    (104, 1, "2026-09-04"),
    (105, 4, "2026-09-05")
]

order_items = [
    (1, 101, 1, 1),
    (2, 101, 2, 2),
    (3, 102, 4, 1),
    (4, 103, 5, 5),
    (5, 104, 3, 1),
    (6, 105, 4, 2)
]

conn.executemany("INSERT OR IGNORE INTO customers VALUES (?, ?, ?, ?)", customers)
conn.executemany("INSERT OR IGNORE INTO products VALUES (?, ?, ?, ?)", products)
conn.executemany("INSERT OR IGNORE INTO orders VALUES (?, ?, ?)", orders)
conn.executemany("INSERT OR IGNORE INTO order_items VALUES (?, ?, ?, ?)", order_items)

conn.commit()

print("Sample data inserted successfully!")
```

```python
pd.read_sql("""
SELECT product_name, price
FROM products
ORDER BY price DESC;
""", conn)
```

### Count

```python
pd.read_sql("""
SELECT MIN(price) AS minimum_price
FROM products;
""", conn)
```

### GROUP BY

```python
pd.read_sql("""
SELECT city, COUNT(*) AS customer_count
FROM customers
GROUP BY city;
""", conn)
```

### INNER JOIN

```python
pd.read_sql("""
SELECT
    customers.customer_name,
    orders.order_id,
    orders.order_date
FROM customers
INNER JOIN orders
ON customers.customer_id = orders.customer_id;
""", conn)
```

### LEFT JOIN

```python
pd.read_sql("""
SELECT
    customers.customer_name,
    orders.order_id
FROM customers
LEFT JOIN orders
ON customers.customer_id = orders.customer_id;
""", conn)
```

### RIGHT JOIN 

```python
pd.read_sql("""
SELECT
    customers.customer_name,
    orders.order_id
FROM customers
RIGHT JOIN orders
ON customers.customer_id = orders.customer_id;
""", conn)
```

## Business Questions
Here are the 10 business question with SQL queries answer about the customers, products, orders and the sales

## 1. How many Customers arre there?

```python
pd.read_sql("""
SELECT COUNT(*) AS total_customers
FROM customers;
""", conn)
```

## 2. How many customers are in each city?

```python
pd.read_sql("""
SELECT city, COUNT(*) AS customers
FROM customers
GROUP BY city;
""", conn)
```

## 3. What is the average customer age?

```python
pd.read_sql("""
SELECT AVG(age) AS average_age
FROM customers;
""", conn)
```

## 4. What is the most expensive product?

```python
pd.read_sql("""
SELECT product_name, price
FROM products
ORDER BY price DESC
LIMIT 1;
""", conn)
```

## 5. What is the cheapest product?

```python
pd.read_sql("""
SELECT product_name, price
FROM products
ORDER BY price ASC
LIMIT 1;
""", conn)
```

## 6. How many products are in each category?

```python
pd.read_sql("""
SELECT category, COUNT(*) AS product_count
FROM products
GROUP BY category;
""", conn)
```

## 7. How many orders has each customer placed?

```python
pd.read_sql("""
SELECT
    customers.customer_name,
    COUNT(orders.order_id) AS total_orders
FROM customers
LEFT JOIN orders
ON customers.customer_id = orders.customer_id
GROUP BY customers.customer_name;
""", conn)
```

## 8. Which products were ordered?

```python
pd.read_sql("""
SELECT
    products.product_name,
    SUM(order_items.quantity) AS quantity_sold
FROM products
INNER JOIN order_items
ON products.product_id = order_items.product_id
GROUP BY products.product_name;
""", conn)
```

## 9. Which customer bought which product?

```python
pd.read_sql("""
SELECT
    customers.customer_name,
    products.product_name,
    order_items.quantity
FROM customers
JOIN orders
ON customers.customer_id = orders.customer_id
JOIN order_items
ON orders.order_id = order_items.order_id
JOIN products
ON order_items.product_id = products.product_id;
""", conn)
```

## 10. What is the total sales value for each product?

```python
pd.read_sql("""
SELECT
    products.product_name,
    SUM(products.price * order_items.quantity) AS total_sales
FROM products
JOIN order_items
ON products.product_id = order_items.product_id
GROUP BY products.product_name
ORDER BY total_sales DESC;
""", conn)
```
