# Exception Handling
Exception handling prevents a program from stopping unexpectedly
when an error occurs.

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print("Result:", result)

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("Zero cannot be used as the divisor.")
```

# Basic Object-Oriented Programming 
Object-oriented programming allows us to organize programs using classes and objects.


```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display_product(self):
        print("Product:", self.name)
        print("Price:", self.price)


product = Product("Running Shoes", 2500)

product.display_product()
```

# Exercise 18: Count Product Categories

## Objective
Count how many times each product category appears.

```python
categories = [
    "Shoes",
    "Clothing",
    "Shoes",
    "Accessories",
    "Shoes",
    "Clothing"
]

category_count = {}

for category in categories:
    if category in category_count:
        category_count[category] += 1
    else:
        category_count[category] = 1

print("Category counts:")

for category, count in category_count.items():
    print(category, ":", count)
```

# Exercise 19: Sales Summary

## Objective
Create a function that calculates total, average, highest and lowest sales.

```python
def sales_summary(sales):
    total = sum(sales)
    average = total / len(sales)
    highest = max(sales)
    lowest = min(sales)

    return total, average, highest, lowest


sales = [1200, 2500, 1800, 3200, 2100]

total, average, highest, lowest = sales_summary(sales)

print("Total Sales:", total)
print("Average Sales:", average)
print("Highest Sale:", highest)
print("Lowest Sale:", lowest)
```
