## 1. NumPy Fundamentals

NumPy provides arrays and numerical operations that are useful for working with numerical data.
In this section, array creation, indexing, broadcasting, and vectorized operations will be practiced.

```python
numbers = np.array([10, 20, 30, 40, 50])

print("NumPy array:", numbers)
print("Array type:", type(numbers))
```

### NumPy Array Indexing

Indexing is used to access individual elements from a NumPy array.
Python uses zero-based indexing, so the first element has index 0.

```python
print("First element:", numbers[0])
print("Third element:", numbers[2])
print("Last element:", numbers[-1])
```

### NumPy Broadcasting

Broadcasting allows a value to be applied to every element of a NumPy array without writing a separate loop.

```python
increased_numbers = numbers + 5

print("Original array:", numbers)
print("After adding 5:", increased_numbers)
```

### NumPy Vectorized Operations

Vectorized operations allow calculations to be performed on all elements of an array at the same time.

```python
prices = np.array([100, 200, 300, 400, 500])

discounted_prices = prices * 0.90

print("Original prices:", prices)
print("Prices after 10% discount:", discounted_prices)
```

## Checking Missing Values

Missing values are checked before cleaning the dataset.

Identifying missing values helps determine which columns require data cleaning or treatment.

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

### Missing Value Percentage

The percentage of missing values in each column is calculated to understand the extent of missing data.

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing value percentage:")
print(missing_percentage)
```

## Data Cleaning

The raw dataset is kept unchanged, and a copy is created for the cleaning process.The missing values in the dataset will be handled using appropriate methods based on the type and amount of missing data.

```python
cleaned_df = df.copy()

print("Original dataset shape:", df.shape)
print("Cleaning dataset shape:", cleaned_df.shape)
```

### Handling Missing Age Values

The Age column contains missing numerical values.The median age is used to fill the missing values because the median is less affected by extreme values than the mean.

```python
age_median = cleaned_df["Age"].median()

cleaned_df["Age"] = cleaned_df["Age"].fillna(age_median)

print("Median age used:", age_median)
print("Missing Age values after cleaning:", cleaned_df["Age"].isnull().sum())
```

### Handling Missing Embarked Values

The Embarked column is categorical and contains a small number of missing values.The missing values are replaced with the most frequently occurring value, known as the mode.

```python
embarked_mode = cleaned_df["Embarked"].mode()[0]

cleaned_df["Embarked"] = cleaned_df["Embarked"].fillna(embarked_mode)

print("Mode used:", embarked_mode)
print("Missing Embarked values after cleaning:", cleaned_df["Embarked"].isnull().sum())
```

### Handling the Cabin Column

The Cabin column contains approximately 77% missing values. Because most records do not contain cabin information, filling such a large proportion of the column could introduce unreliable information. Therefore, the Cabin column is removed from the cleaned analysis-ready dataset.

```python
cleaned_df = cleaned_df.drop(columns=["Cabin"])

print("Cabin column removed.")
```

### Checking Missing Values After Cleaning

The dataset is checked again to confirm that the missing values have been handled.

```python
print("Missing values after cleaning:")
print(cleaned_df.isnull().sum())
```

## Checking and Removing Duplicate Records

Duplicate records can cause the same information to be counted more than once. The dataset is checked for duplicate rows before removing them.

```python
duplicate_count = cleaned_df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

## Filtering Data

Filtering is used to select specific rows from a dataset based on a condition. In this section, passengers from a particular passenger class will be selected from the cleaned dataset.

```python
first_class = cleaned_df[cleaned_df["Pclass"] == 1]

print("Number of first-class passengers:", len(first_class))

first_class.head()
```

### Filtering Passengers by Survival Status

The dataset can also be filtered to select passengers who survived.

```python
survived_passengers = cleaned_df[cleaned_df["Survived"] == 1]

print("Number of passengers who survived:", len(survived_passengers))

survived_passengers.head()
```

### Filtering Using Multiple Conditions

Multiple conditions can be combined to obtain more specific records. The following example selects female passengers who were in first class.

```python
female_first_class = cleaned_df[
    (cleaned_df["Sex"] == "female") &
    (cleaned_df["Pclass"] == 1)
]

print("Number of female first-class passengers:", len(female_first_class))

female_first_class.head()
```

##  Grouping Data Using groupby()

The groupby() function is used to divide data into groups based on a particular column and calculate summary information for each group. In this section, passenger survival will be summarized by passenger class.

```python
survival_by_class = cleaned_df.groupby("Pclass")["Survived"].mean()

print("Average survival rate by passenger class:")
print(survival_by_class)
```

### Grouping Survival by Sex

The survival rate can also be grouped by sex to compare the values between male and female passengers.

```python
survival_by_sex = cleaned_df.groupby("Sex")["Survived"].mean()

print("Average survival rate by sex:")
print(survival_by_sex)
```

### Grouping by Multiple Columns
Pandas can group data using more than one column. The following example calculates the average survival rate for each combination of passenger class and sex.

```python
survival_by_class_and_sex = cleaned_df.groupby(
    ["Pclass", "Sex"]
)["Survived"].mean()

print(survival_by_class_and_sex)
```

## Data Transformation

Data transformation means creating or modifying columns so that the dataset becomes more useful for analysis. In this section, we will create new features from the existing Titanic dataset: `Family_Size`, `Is_Alone`,`Fare_Per_Person`. These transformations make the dataset more suitable for analysis.


```python
cleaned_df["Family_Size"] = cleaned_df["SibSp"] + cleaned_df["Parch"] + 1

cleaned_df[["SibSp", "Parch", "Family_Size"]].head()
```

### Identifying Passengers Travelling Alone

The `Is_Alone` column identifies whether a passenger was travelling alone. A passenger is considered alone when `Family_Size` is equal to 1.

```python
cleaned_df["Is_Alone"] = cleaned_df["Family_Size"] == 1

cleaned_df[["Family_Size", "Is_Alone"]].head()
```

### Calculating Fare Per Person

The `Fare_Per_Person` column calculates the fare relative to the passenger's family size. This creates another useful numerical feature for analysis.

```python
cleaned_df["Fare_Per_Person"] = (
    cleaned_df["Fare"] / cleaned_df["Family_Size"]
)

cleaned_df[["Fare", "Family_Size", "Fare_Per_Person"]].head()
```

### Check the Transformed Dataset

The following code displays the new columns and confirms that the transformations were applied successfully.

```python
cleaned_df[
    ["PassengerId", "Pclass", "Fare", "Family_Size",
     "Is_Alone", "Fare_Per_Person"]
].head(10)
```

## Merging Datasets

Merging allows information from separate DataFrames to be combined using a common column. To demonstrate merging, we create a small lookup DataFrame containing the passenger class number and its corresponding class name. We then merge this information into the cleaned Titanic dataset using `Pclass`.

```python
class_info = pd.DataFrame({
    "Pclass": [1, 2, 3],
    "Class_Name": ["First Class", "Second Class", "Third Class"]
})

class_info
```

### Merging 

```python
cleaned_df = cleaned_df.merge(
    class_info,
    on="Pclass",
    how="left"
)

cleaned_df.head()
```
